# Algorithm

Find repeating patterns

Need a function that can tell whether number is or not valid

```python
def is_valid(number):
    snumber = str(number)

    #check by size, maximum is half len

    mid = len(snumber)//2
    
    s0,s1 = snumber[:mid] , snumber[mid:]
            #print(f"{s0=}, {s1=}")

    if s0==s1:
        return False
    
    return True
```

```python
print(is_valid(11))
print(is_valid(22))
print(is_valid(2020))
print(is_valid(12343456))
print(is_valid(1188511885))

print(is_valid(123456))
```

```python
def get_invalids_in_range(start,end):
    # includes end
    invs = []
    for n in range(start, end+1):
        if not is_valid(n):
            print("invalid number:", n)
            invs.append(n)
    
    return invs
```

```python
import sympy

def is_invalid_sub(number):
    # get divisors of string length
    snumber = str(number)

    size_divisors = sympy.divisors(len(snumber))

    for size in size_divisors[:-1]:
        # get first element
        s0 = snumber[0:size]
        
        is_repeating=True
        # check against all the others
        for i in range(size, len(snumber), size):
            s1 = snumber[i:i+size]

            if s0!=s1:
                is_repeating=False
                break
        
        if is_repeating: return True
    
    return False
```

```python
is_invalid_sub(2121212121)
```

```python
is_invalid_sub(1698528)
```

```python
is_invalid_sub(1698522)
```

```python
def get_invalids_sub_in_range(start,end):
    # includes end
    invs = []
    for n in range(start, end+1):
        if is_invalid_sub(n):
            print("invalid number:", n)
            invs.append(n)
    
    return invs
```

```python
sum_inv_subs=0
for rstart, rend in ranges:
    invs0 = get_invalids_sub_in_range(rstart,rend)
    sum_inv_subs += sum(invs0)
```
