```python
TEST = False

INPUT_FN = "input_test.txt"
TEST_SOLUTION = 3

if not TEST:
    INPUT_FN = "input.txt"
```

```python
# with open(INPUT_FN) as f:
#     s = f.read()

# patterns_block, designs_block = s.split('\n\n')

# Read and split lines
with open(INPUT_FN) as f:
    lines = [l.rstrip() for l in f]

if lines[-1]=='':
    lines=lines[:,-1]
```

```python
lines
```

Algorithm

```python
positions = [] # record end positions
count_end_zeros = 0

position = 50

for l0 in lines:
    dir0 = l0[0]

    moves = int(l0[1:])

    if dir0=='L':
        moves = -moves

    print(f"{l0=}, {position=}, {dir0=}")
    position += moves

    position = position % 100

    print(f"new position {position}")

    positions.append(position)
    if position==0: count_end_zeros+=1

print(f"{count_end_zeros=}")
```
