Suggested algorithm. To keep track of the row location of tachyon beams in a set. Run a for loop until end.

```python
nrows = len(lines[0])
nrows
```

```python
tachyons_rows = set()
count_splits=0

for i, line0 in enumerate(lines):
    printd(f"{i=}, {line0=}")
    if i == 0:
        # set first tachyon
        loc = line0.find('S')
        printd(f"starting location row {loc}")
        tachyons_rows.add(loc)
    elif i == len(lines)-1: #last line exits
        break
    else:
        tach_rows0 = set() # build a new list

        for i_tach in tachyons_rows:
            #printd(f"tachyon {i_tach}")
            # check if there is a splitter at the same location
            if line0[i_tach]=='^':
                #splitter
                count_splits+=1
                if i_tach>0:
                    tach_rows0.add(i_tach-1)
                if i_tach<nrows-1:
                    tach_rows0.add(i_tach+1)

            else:
                tach_rows0.add(i_tach)
        
        tachyons_rows = tach_rows0 #replace with new locations of tachyons
```

```python
if TEST:
    assert TEST_SOLUTION == count_splits
count_splits
```

# Part 2

Count the number of timelines

Need to keep track of the number or routes taken to reach each point

```python
from collections import defaultdict

tachyons_rows_count = None

for i, line0 in enumerate(lines):
    printd(f"{i=}, {line0=}")
    if i == 0:
        # set first tachyon
        loc = line0.find('S')
        printd(f"starting location row {loc}")
        
        tachyons_rows_count = defaultdict(int)
        tachyons_rows_count[loc] = 1

    elif i == len(lines)-1: #last line exits
        break
    else:
        tachyons_rows_count0 = defaultdict(int)
        for tach_icol, count in tachyons_rows_count.items():
            #consider only tachyons in the line being considered
            if line0[tach_icol]=='^':
                #splitter
                if tach_icol>0:
                    tachyons_rows_count0[tach_icol-1] += count
                if tach_icol<nrows-1:
                    tachyons_rows_count0[tach_icol+1] += count

            else:
                tachyons_rows_count0[tach_icol] += count
        tachyons_rows_count = tachyons_rows_count0
```

```python
tachyons_rows_count
```

```python
count_timelines = sum(tachyons_rows_count.values())
```

```python
if TEST:
    assert TEST_SOLUTION2 == count_timelines
count_timelines
```
