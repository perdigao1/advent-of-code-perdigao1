# Parse shapes and regions

```python
shapes = []
regions = []

line_iter = iter(lines)

for l0 in line_iter:
    if ':' in l0:
        if not 'x' in l0:
            # shape
            # get number
            number = int(l0[0])
            shape=[]
            area = 0
            # get next 3 lines
            for _ in range(3):
                sh = next(line_iter)
                # count number of #
                area+= sh.count('#')
                shape.append(sh)
            shapes.append( { 
                'shape':shape,
                'area': area
                })
        else:
            # region
            size , qties = l0.split(':')
            
            size0,size1 = map(int, size.split('x'))

            qties = qties.strip()
            values = list(map(int, qties.split(' ')))

            regions.append({'size':(size0,size1),
                            'values': values})
```

# Try fit using simple rules of not fit or fit

```python
count_valid = 0
for reg0 in regions:
    print(f"region {reg0}")
    s = reg0['size']
    values = reg0['values']
    
    # check if area is large enough
    region_area = s[0]*s[1]

    all_blocks_area=0
    count_blocks = 0
    for v0, sh0 in zip(values, shapes):
        all_blocks_area+= v0* sh0['area']
        count_blocks+= v0

    print(f"{region_area=} , {all_blocks_area=}")
    if region_area<all_blocks_area:
        print("Not possible to fit")
        continue

    # if it is possible to fit check how many 3x3 areas
    # it contains and check total number of shapes
    area_3_3 = (s[0]//3) * (s[1]//3)
    print(f"{area_3_3=}, { count_blocks=}")
    if count_blocks<=area_3_3:
        print("Can fit blocks")
        count_valid+=1
        continue
    
    print("NOT SURE")
```
