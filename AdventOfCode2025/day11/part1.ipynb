Too many nodes, exponential, takes too long. Try DP

```python
def count_simple_paths_dag(G, source, target):
    # Topological sort
    order = list(nx.topological_sort(G))
    # DP table: dp[node] = number of paths from source to node
    dp = {node: 0 for node in G}
    dp[source] = 1

    for u in order:
        for v in G.successors(u):
            dp[v] += dp[u]

    return dp[target]
```

```python
count_routes=0
for r0 in routes:
    print(f"{r0=}")
    nroutes = 1
    for i in range(len(r0)-1):
        source=r0[i]
        target=r0[i+1]

        others = r0.copy()
        others.remove(source)
        others.remove(target)

        printd(f"{source=}, {target=}, {others=}")

        # check routes from source to target with others removed
        G1 = G.copy()
        #printd(G1.nodes)
        for o0 in others:
            G1.remove_node(o0)
        
        nroutes1 = count_simple_paths_dag(G1, source, target)
        nroutes*=nroutes1
    printd(f"route {r0=}, has {nroutes=}")
    count_routes+=nroutes
```

```python
solution = count_routes
```
