# Algorithm

Just count the rolls with counts. Iterate through all the rolls

```python
nrows = len(lines)
ncols = len(lines[0])
```

```python
import numpy as np
counter = np.zeros( (nrows,ncols), dtype=np.uint32 )
mask = np.zeros_like(counter, dtype=bool)
```

```python
mask
```

```python
roll_neighbours_count = np.ma.array( counter, mask=mask)
```

```python
roll_neighbours_count.mask
```

```python
for irow,line in enumerate(lines):
    for icol,c in enumerate(line):
        if c == '@':
            #check neighbours
            dirs = []
            if irow>0:
                dirs.append((-1,0))
                if icol>0:
                    dirs.append((-1,-1))
                if icol<ncols-1:
                    dirs.append((-1,1))
            if irow<nrows-1:
                dirs.append((1,0))
                if icol>0:
                    dirs.append((1,-1))
                if icol<ncols-1:
                    dirs.append((1,1))
            if icol>0:
                dirs.append((0,-1))
            if icol<ncols-1:
                dirs.append((0,1))

            for dir in dirs:
                irow1, icol1 = ( irow+dir[0], icol+dir[1] )

                c1 = lines[irow1][icol1]
                if c1 == '@':
                    roll_neighbours_count[irow,icol]+=1
        else:
            roll_neighbours_count.mask[irow,icol]=True
```

```python
roll_neighbours_count
```

```python
access_rolls = roll_neighbours_count<4
access_rolls
```

```python
result = np.sum(access_rolls)
```

```python
if TEST:
    assert result== TEST_SOLUTION

print(int(result))
```

# Part2

Remove, remove, until stop

Convert map to boolean grid

Mark to be removed

Remove at next iteration

```python
rolls_map = np.zeros_like(counter, dtype=bool)

for irow,line in enumerate(lines):
    for icol,c in enumerate(line):
        if c=='@':
            rolls_map[irow,icol] = True
        else:
            rolls_map[irow,icol]=False
```

```python
rolls_map
```

Rewrite the algorithm to pick rolls to remove and then remove iteratively

```python
rolls_map0 = rolls_map.copy()
count0 = np.sum(rolls_map0)
print(f"{count0}")

while True:
    counter = np.zeros_like( rolls_map0 , dtype=np.uint32 )
    mask = np.zeros_like(rolls_map0, dtype=bool)

    roll_neighbours_count = np.ma.array( counter, mask=mask)

    roll_neighbours_count
    for irow in range(nrows):
        for icol in range(ncols):
            if rolls_map0[irow,icol]:
                #check neighbours
                dirs = []
                if irow>0:
                    dirs.append((-1,0))
                    if icol>0:
                        dirs.append((-1,-1))
                    if icol<ncols-1:
                        dirs.append((-1,1))
                if irow<nrows-1:
                    dirs.append((1,0))
                    if icol>0:
                        dirs.append((1,-1))
                    if icol<ncols-1:
                        dirs.append((1,1))
                if icol>0:
                    dirs.append((0,-1))
                if icol<ncols-1:
                    dirs.append((0,1))

                for dir in dirs:
                    irow1, icol1 = ( irow+dir[0], icol+dir[1] )

                    if rolls_map0[irow1,icol1]:
                        roll_neighbours_count[irow,icol]+=1
            else:
                roll_neighbours_count.mask[irow,icol]=True
    
    access_rolls = roll_neighbours_count<4
    # here the rolls to be removed will be marked
    sum_access_rolls  = np.sum(access_rolls)
    print(f"{sum_access_rolls=}")
    if sum_access_rolls==0:
        print("No more rolls to remove. Exiting")
        break

    # remove them

    rolls_map1 = np.logical_not( np.logical_and(rolls_map0, access_rolls))

    rolls_map0 = rolls_map1 #next iteration

count1= np.sum(rolls_map1)
```

```python
rolls_removed = count0-count1
```
