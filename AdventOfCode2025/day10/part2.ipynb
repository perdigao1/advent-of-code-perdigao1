```python
import numpy as np
import pulp

def solve_for_minimal_integer_combination(vectors, target_vector):
    """
    Finds the non-negative integer coefficients for a linear combination
    of the given vectors that equals the target vector, minimizing the sum
    of the coefficients.

    Args:
        vectors (list or np.array): A list of N-dimensional vectors (e.g., [[v1], [v2], ...]).
                                    Shape should be (num_vectors, dimension).
        target_vector (list or np.array): The N-dimensional target vector.
    
    Returns:
        tuple: A tuple containing:
            - np.array: The optimal coefficients (x) if a solution is found, otherwise None.
            - str: A status message from the solver.
    """
    # Convert inputs to numpy arrays for consistent handling
    A_matrix = np.array(vectors).T  # Columns of A are the input vectors
    y_target = np.array(target_vector)

    num_vectors = A_matrix.shape[1]
    dimension = A_matrix.shape[0]

    # Define the ILP problem
    prob = pulp.LpProblem("Minimize_Sum_of_Coefficients", pulp.LpMinimize)

    # Decision variables: the coefficients x_i, constrained to be non-negative integers
    x_vars = [
        pulp.LpVariable(f"x_{i}", lowBound=0, cat=pulp.LpInteger)
        for i in range(num_vectors)
    ]

    # Objective function: minimize the sum of coefficients
    prob += pulp.lpSum(x_vars), "Total_Sum_of_Coefficients"

    # Constraints: A @ x == y
    for dim_idx in range(dimension):
        lhs_expr = pulp.lpSum(A_matrix[dim_idx, j] * x_vars[j] for j in range(num_vectors))
        prob += lhs_expr == y_target[dim_idx], f"Constraint_Equality_Dim_{dim_idx}"

    # Solve the problem using the default solver
    printd("Solving the integer linear program...")
    prob.solve(pulp.PULP_CBC_CMD(msg=0))  # msg=0 suppresses solver output

    status = pulp.LpStatus[prob.status]

    if pulp.LpStatus[prob.status] == "Optimal":
        solution = np.array([pulp.value(var) for var in x_vars])
        return solution, status
    else:
        return None, status
```
