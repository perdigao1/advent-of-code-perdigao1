Deal with buttons in binary. Pressing a button is a XOR operation

```python
machines_bin = []

for l0 in lines:
    split_line = l0.split(" ")
    indicator_str = split_line[0][1:-1]
    indic_bin_str = ''.join(['1' if c1=='#' else '0' for c1 in indicator_str])[::-1]
    indic_int = int(indic_bin_str,2)
    printd(f"{indicator_str=}, {indic_bin_str=}, {indic_int=}")

    wiring_schemes=[]
    for ws0 in split_line[1:-1]:
        values_str_split = ws0[1:-1].split(',')
        values = list(map(int, values_str_split))
        wr_scheme = 0
        for v in values:
            wr_scheme+= 2**v
        wiring_schemes.append(wr_scheme)

    jolt_str = split_line[-1]

    machines_bin.append({'indicator': indic_int,
                     'wiring_schemes': wiring_schemes,
                     'jolt_str': jolt_str})
```

We should treat the wiring schemes as some sort of boolean when iterating through elements

```python
import itertools
def binary_combinations(lst):
    """Yield combinations of elements from lst based on binary counter."""
    n = len(lst)
    for bits in itertools.product([0, 1], repeat=n):
        # Pick elements where bit == 1
        yield [lst[i] for i in range(n) if bits[i] == 1]
```

Find the shortest combination 

It is probably worthwhile just looking at the XOR of all elements in each line of `bin_combin`. If it matches the value desired, we can keep it, and get the length of the array as number of pushes

```python
min_pushes=None
ways = []

for binc0 in bin_combin:
    v = 0 # starting value
    for i0 in binc0:
        v ^= i0
    
    if v==machines_bin[0]['indicator']:
        ways.append(binc0)
        
        #min
        if min_pushes is None:
            min_pushes=len(binc0)
        else:
            min_pushes=min(min_pushes, len(binc0))
```

```python
min_pushed_sum=0

for machine0 in machines_bin:
     
    bin_combin0 = list(binary_combinations( machine0['wiring_schemes']))
    min_pushes=None


    for binc0 in bin_combin0:
        v = 0 # starting value
        for i0 in binc0:
            v ^= i0
        
        if v==machine0['indicator']:
            
            #min
            if min_pushes is None:
                min_pushes=len(binc0)
            else:
                min_pushes=min(min_pushes, len(binc0))
    
    if min_pushes is not None:
        min_pushed_sum+=min_pushes
```
