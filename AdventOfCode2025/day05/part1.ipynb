```python
from itertools import combinations
```

```python
def clean_up_ranges(ranges):
    ranges0 = ranges.copy()

    has_changed=True
    iter=0
    while has_changed:
        printd(f"{iter=}")
        has_changed=False

        for r0,r1 in combinations(ranges0,2):
            printd(f"Checking {r0=}, {r1=}")
            # is there overlap?

            if (r1[0]>=r0[0] and r1[0]<=r0[1]) or (r1[1]>=r0[0] and r1[1]<=r0[1]) or \
                (r0[0]>=r1[0] and r0[0]<=r1[1]) or (r0[1]>=r1[0] and r0[1]<=r1[1]):

                printd(f"overlap {r0=} , {r1=}")
                # overlap
                has_changed=True
                # create new range

                rmin = min(r0[0], r1[0])
                rmax = max(r0[1], r1[1])
                newrange = (rmin,rmax)

                ranges0.remove(r0)
                ranges0.remove(r1)
                ranges0.append(newrange)
                break
        iter+=1
    
    printd(f"Completed at {iter=}")
    return ranges0
```
