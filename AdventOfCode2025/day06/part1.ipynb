```python
lines
```

seperate numbers and operations

```python
numbers_rows=[]
operations=[]

for l0 in lines:
    if '+' in l0 or '*' in l0:
        # operations
        operations = l0.split()

    elif len(l0)!=0:
        numbers_row = list(map(int, l0.split(),)) # splits multispaces as one
        numbers_rows.append(numbers_row)
```

```python
numbers_rows
```

```python
assert len(numbers_row)==len(operations)
```

# Algorithm

```python
answers = []

for i, op0 in enumerate(operations):

    op_numbers = [ nrow[i] for nrow in numbers_rows]
    res = None
    match op0:
        
        case '+':
            res = sum(op_numbers)
        case '*':
            res=1
            for n0 in op_numbers:
                res*=n0
    answers.append(res)
answers
```

```python
grand_total = sum(answers)
```

```python
if TEST:
    assert grand_total == TEST_SOLUTION
grand_total
```
