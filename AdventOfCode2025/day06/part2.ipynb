The location of the operation characters mark the start and end of the digits to operate. So don't split, but rather do a character search.

```python
op_chars_indices = []
nops=0
for i,c in enumerate(op_str):
    if c=='+' or c=='*':
        op_chars_indices.append(i)
        nops+=1

#op_chars_indices.append(len(op_str))
```

```python
op_chars_indices
```

```python
nops
```

These specify the range of cols that the search will take place

```python
results = []
for iop in range(nops):
    istart = op_chars_indices[iop]
    iend = None
    if iop==nops-1:
        iend = len(op_str)
    else:
        iend = op_chars_indices[iop+1]
    print(f"{istart=}, {iend=}")

    col_values = []
    for icol in range(istart,iend):
        col_str = ""
        
        for nrow_str in numbers_strs:
            col_str+= nrow_str[icol]
        try:
            col_value = int(col_str)
            col_values.append(col_value)
            printd(f"{col_value=}")
        except ValueError as e:
            pass
    

    
    # apply operation
    op = op_str[istart]
    printd(f"{op=}")
    res = 0
    match op:
        case '+':
            res = sum(col_values)
        case '*':
            res=1
            for n0 in col_values:
                res*=n0
    
    printd(f"{res=}\n")
    results.append(res)
```

```python
grand_total = sum(results)
```
