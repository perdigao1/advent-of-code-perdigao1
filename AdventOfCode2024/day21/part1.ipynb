```python
class cRobot_Base:
    robot_keyp_klocs = None
    cur_loc = 'A'
        
    def get_and_make_move(self,loc):
        ret = self.get_moves_to(loc)
        self.cur_loc=loc

        return ret
    def reset(self):
        self.cur_loc='A'

    def do_moves(self,smoves):
        #useful for simulation
        cur_coords = self.robot_keyp_klocs[self.cur_loc]
        new_coords = cur_coords
        
        for c in smoves:
            match c:
                case '^':
                    new_coords= ( new_coords[0]-1, new_coords[1])
                case 'v':
                    new_coords = ( new_coords[0]+1, new_coords[1])
                case '>':
                    new_coords = ( new_coords[0], new_coords[1]+1)
                case '<':
                    new_coords = ( new_coords[0], new_coords[1]-1)
        printd(f"{cur_coords=}, {new_coords=}")

        #update location string
        for k,v in self.robot_keyp_klocs.items():
            if v==new_coords:
                printd(f"new position {k}")
                self.cur_loc = k

        return self.cur_loc
```

```python
class cRobot_num_keypad(cRobot_Base):
    robot_keyp_klocs = {
        '7':(0,0), '8':(0,1), '9':(0,2),
        '4':(1,0), '5':(1,1), '6': (1,2),
        '1':(2,0), '2':(2,1), '3':(2,2),
                '0':(3,1), 'A':(3,2)
    }
    
    last_move = None

    def get_moves_to(self,loc):
        if isinstance(loc,int):
            loc=str(loc)


        p0 = self.robot_keyp_klocs[self.cur_loc]
        p1 = self.robot_keyp_klocs[loc]

        dr = p1[0]-p0[0]
        dc = p1[1]-p0[1]
        printd(f"{self.cur_loc=}, {loc=}, {dr=},{dc=}")

        moves=""

        priority_r=False
        movesr=""
        if dr<0:
            movesr="^"*-dr
            if self.last_move=='^':
                priority_r=True
        elif dr>0:
            movesr="v"*dr
            if self.last_move=='v':
                priority_r=True
        
        priority_c = False
        movesc=""
        if dc>0:
            movesc=">"*dc
            if self.last_move==">":
                priority_c=True
        elif dc<0:
            movesc="<"*-dc
            if self.last_move=="<":
                priority_c=True

        moves = movesc + movesr # default

        if priority_r:
            moves = movesr + movesc
        if priority_c:
            moves= movesc + movesr
        
        if self.cur_loc=='0' and dc<0:
            moves = movesr + movesc
        
        if loc=='0' and dc>0:
            moves = movesc + movesr
        
        self.last_move= moves[-1]

        return moves
```

```python
class cRobot_dir_keypad(cRobot_Base):
    robot_keyp_klocs = {
                   '^':(0,1), 'A':(0,2),
        '<':(1,0), 'v':(1,1), '>': (1,2)
    }

    def get_moves_to(self,loc):

        p0 = self.robot_keyp_klocs[self.cur_loc]
        p1 = self.robot_keyp_klocs[loc]

        dr = p1[0]-p0[0]
        dc = p1[1]-p0[1]
        printd(f"{self.cur_loc=}, {loc=}, {dr=},{dc=}")

        moves=""

        if loc=="<":
            #move down first
            if dr>0:
                moves +="v"*dr

            #then left
            if dc<0:
                moves+="<"*-dc

            return moves
        
        if dc>0:
            moves+=">"*dc
        elif dc<0:
            moves+="<"*-dc

        if dr<0:
            moves+="^"*-dr
        elif dr>0:
            moves+="v"*dr
            
        return moves
```
